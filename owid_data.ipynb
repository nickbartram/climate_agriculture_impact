```python
import pandas as pd
import plotly.express as px
```

```python
# Load Our World in Data CO2 data ('owid-co2-data.csv')
owid_co2_df = pd.read_csv('Resources/owid-co2-data.csv')
owid_co2_df.head()
```

```python
# Look at year range for entire notebook
year_min = owid_co2_df['year'].min()
year_max = owid_co2_df['year'].max()

# Print statement of min/max years
print(f"The earliest year recorded in this OWID CO2 dataset: {year_min}. ")
print(f"The latest year recorded in this OWID CO2 dataset:   {year_max}. ")
```

```python
# Look at unique countries in the dataset
country_list = owid_co2_df['country'].unique()

country_sum = owid_co2_df['country'].nunique()

print(f"There are {country_sum} countries in this dataset.")
print(f"The following countries are represented in the dataset: {country_list}")
```

For the purpose of this project we only want to see results from Canada and the United States, and only frmo 1908 to present day.

```python
# Filter dataset for only Canada and the United States
filtered_df = owid_co2_df[owid_co2_df['country'].isin(['Canada', 'United States'])]

remaining_countries = filtered_df['country'].unique()

print(f"The remainging countries after filtering are: {remaining_countries}")
```

```python
# Look at min/max range of year in filtered_df
filtered_year_min = filtered_df['year'].min()
filtered_year_max = filtered_df['year'].max()

print(f"The earliest year in the filtered Canada and USA dataset: {filtered_year_min}")
print(f"The latest year in the filtered Canada and USA dataset:   {filtered_year_max}")
```

```python
# Filter out dates before 1908
post_1908_df = filtered_df[filtered_df['year'] >= 1908]
post_1908_df.head()
```

Some good work here eliminating all but Canada and the USA, and also extra years before 1908. Dates before 1908 were eliminated to line up with crop yield datasets to come later (Stats Canada and USDA).

There are still a ton of columns here, some can definitely be dropped.

```python
# Look at the columns
post_1908_df.columns
```

```python
post_1908_reduced_df = post_1908_df[['country', 'year', 'iso_code', 'population', 'gdp',
                                               'co2', 'co2_growth_prct', 'co2_per_capita', 
                                               'co2_per_gdp', 'total_ghg']]

post_1908_reduced_df.head()
```

```python
# Look for Nan entries
post_1908_reduced_df.isnull().sum()
```

```python
# Look for NaN rows
nan_rows = post_1908_reduced_df[post_1908_reduced_df.isnull().any(axis=1)]
nan_rows
```

Encouragingly there are only 4 total NaN entries in this dataset. They both refer to GDP in 2023, presumably the dataset was made before these numbers were released, or at least they were unavailable at that time. GPD is not vital to this project, but for now we will keep this row, the rest of the data is important.

```python
# Let's plot c02 levels in the North America since 1908 (Canada and United States combined)
# Group by year and sum CO₂ across all countries
NA_co2_per_year = post_1908_reduced_df.groupby('year', as_index=False)['co2'].sum()

# Plot with Plotly Express
fig = px.line(NA_co2_per_year, 
              x='year', 
              y='co2',
              title='Total CO₂ Emissions per Year (All Countries)',
              labels={'co2': 'Total CO₂ emissions (Mt)', 'year': 'Year'})

fig.show()
```

```python
# Group by year and sum CO₂ of Canada
# Filter for Canada
canada_df = post_1908_reduced_df[post_1908_reduced_df['country'] == "Canada"]

canada_co2_per_year = canada_df.groupby('year', as_index=False)['co2'].sum()

# Plot with Plotly Express
fig = px.line(canada_co2_per_year, 
              x='year', 
              y='co2',
              title='Total CO₂ Emissions per Year (Canada)',
              labels={'co2': 'Total CO₂ emissions (Mt)', 'year': 'Year'})

fig.show()
```

```python
# Group by year and sum CO₂ of USA
usa_df = post_1908_reduced_df[post_1908_reduced_df['country'] == "United States"]

usa_co2_per_year = usa_df.groupby('year', 'country' == "United States", as_index=False)['co2'].sum()

# Plot with Plotly Express
fig = px.line(usa_co2_per_year, 
              x='year', 
              y='co2',
              title='Total CO₂ Emissions per Year (USA)',
              labels={'co2': 'Total CO₂ emissions (Mt)', 'year': 'Year'})

fig.show()
```

```python
# Group both countries
global_co2_per_year = post_1908_reduced_df.groupby(['year', 'country'], as_index=False)['co2'].sum()

# Plot with Plotly Express
fig = px.line(global_co2_per_year, 
              x='year', 
              y='co2',
              color='country',
              title='Total CO₂ Emissions per Year (USA and Canada)',
              labels={'co2': 'Total CO₂ emissions (Mt)', 'year': 'Year'})

fig.show()
```
