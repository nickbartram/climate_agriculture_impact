```python
import pandas as pd
import requests
import plotly.express as px
```

```python
# Get Climate Change Knowledge Portal API dataset about rainfall in North America

# Save url
url = 'https://cckpapi.worldbank.org/cckp/v1/cru-x0.5_timeseries_pr_timeseries_annual_1901-2024_mean_historical_cru_ts4.09_mean/region_nac?_format=json'
```

```python
# Call the API
response = requests.get(url)
response.raise_for_status()
```

```python
# Convert to Python dict
data = response.json()
```

```python
# Extract the nested 'data' dict
nested_data = data['data']
```

```python
# Build a list with data
records = []
for region_code, timeseries in nested_data.items():
    if timeseries is None:
        continue  # skip regions with no data
    for date, value in timeseries.items():
        records.append({
            "region_code": region_code,
            "date": date,
            "rainfall_mm": value
        })
```

```python
# Now try to create a DataFrame
rainfall_df = pd.DataFrame(records)
```

```python
# We need to find only regions in Canada and the USA
north_america_codes = ['CAN', 'USA']  # add more if needed
rainfall_df_na = rainfall_df[rainfall_df['region_code'].str.startswith(tuple(north_america_codes))].copy()
```

```python
# Cast 'date' column from object to datetime
rainfall_df_na['date'] = pd.to_datetime(rainfall_df_na['date'])
```

```python
# View date column
rainfall_df_na['date']
```

```python
# Convert 'date' column to year integer
rainfall_df_na['date'] = rainfall_df_na['date'].dt.year
```

```python
# Filter out dates before 1908
post_1908_rainfall_na_df = rainfall_df_na[rainfall_df_na['date'] >= 1908]
post_1908_rainfall_na_df.head()
```

```python
post_1908_rainfall_na_df.info()
```

```python
# Rename 'date' column to 'year'
post_1908_rainfall_na_df = post_1908_rainfall_na_df.rename(columns={'date': 'year'})
```

```python
# Let's plot rainfall levels in the North America since 1908 (Canada and United States combined)
# Group by year and sum rainfall across all countries
NA_rainfall_per_year = post_1908_rainfall_na_df.groupby('year', as_index=False)['rainfall_mm'].sum()

# Plot with Plotly Express
fig = px.scatter(NA_rainfall_per_year, 
              x='year', 
              y='rainfall_mm',
              trendline= 'ols',
              title='Total Rainfall per Year (All Countries)',
              labels={'rainfall_mm': 'Total Rainfall (mm)', 'year': 'Year'})

# Make trendline red
fig.data[1].line.color = "red"
fig.data[1].line.width = 2 

fig.show()
```

```python
# Let's plot rainfall levels in the North America since 1908 (Canada and United States combined)
# Group by year and sum rainfall across all countries
NA_rainfall_per_year = post_1908_rainfall_na_df.groupby('year', as_index=False)['rainfall_mm'].sum()

# Plot with Plotly Express
fig = px.line(NA_rainfall_per_year, 
              x='year', 
              y='rainfall_mm',
              title='Total Rainfall per Year (All Countries)',
              labels={'rainfall_mm': 'Total Rainfall (mm)', 'year': 'Year'})

fig.show()
```

```python
from sklearn.linear_model import LinearRegression
import plotly.graph_objects as go
import numpy as np

# Prepare data
X = NA_rainfall_per_year['year'].values.reshape(-1, 1)
y = NA_rainfall_per_year['rainfall_mm'].values

# Fit linear regression
model = LinearRegression()
model.fit(X, y)

# Predict values
y_pred = model.predict(X)

# Plot original rainfall as line
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=NA_rainfall_per_year['year'],
    y=y,
    mode='lines',
    name='Rainfall'
))

# Add regression line
fig.add_trace(go.Scatter(
    x=NA_rainfall_per_year['year'],
    y=y_pred,
    mode='lines',
    name='Linear Regression',
    line=dict(dash='dash', color='red')  # dashed red line
))

# Update layout
fig.update_layout(
    title='Total Rainfall per Year (All Countries)',
    xaxis_title='Year',
    yaxis_title='Total Rainfall (mm)'
)

fig.show()
```
