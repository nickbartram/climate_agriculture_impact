```python
import pandas as pd
import requests
import plotly.express as px
from sklearn.linear_model import LinearRegression
import plotly.graph_objects as go
import numpy as np
```

```python
# Get Climate Change Knowledge Portal API dataset about rainfall in North America

# Save url
url = 'https://cckpapi.worldbank.org/cckp/v1/cru-x0.5_timeseries_pr_timeseries_annual_1901-2024_mean_historical_cru_ts4.09_mean/region_nac?_format=json'
```

```python
# Call the API
response = requests.get(url)
response.raise_for_status()
```

```python
# Convert to Python dict
data = response.json()
```

```python
# Extract the nested 'data' dict
nested_data = data['data']
```

```python
# Build a list with data
records = []
for region_code, timeseries in nested_data.items():
    if timeseries is None:
        continue  # skip regions with no data
    for date, value in timeseries.items():
        records.append({
            "region_code": region_code,
            "date": date,
            "rainfall_mm": value
        })
```

```python
# Now try to create a DataFrame
rainfall_df = pd.DataFrame(records)
```

```python
# Check on the df
rainfall_df.head()
```

```python
# Check rainfall_df shape
rainfall_df.shape
```

```python
# Let's look at a list of region codes
list(rainfall_df['region_code'].unique())
```

```python
# Let's look at unique region codes
rainfall_df['region_code'].nunique()
```

```python
# We need to find only regions in Canada and the USA
north_america_codes = ['CAN', 'USA']  # add more if needed
rainfall_df_na = rainfall_df[rainfall_df['region_code'].str.startswith(tuple(north_america_codes))].copy()
```

```python
# Look at new rainfall_df_na
rainfall_df_na.head()
```

```python
# Look at unique region codes
rainfall_df_na['region_code'].nunique()
```

```python
# Look at region codes
rainfall_df_na['region_code'].unique()
```

```python
# Check for na
rainfall_df_na.isna().sum()
```

```python
# Data types
rainfall_df_na.info()
```

```python
# Cast 'date' column from object to datetime
rainfall_df_na['date'] = pd.to_datetime(rainfall_df_na['date'])
```

```python
# View date column
rainfall_df_na['date']
```

```python
# Convert 'date' column to year integer
rainfall_df_na['date'] = rainfall_df_na['date'].dt.year
```

```python
# Filter out dates before 1908
post_1908_rainfall_na_df = rainfall_df_na[rainfall_df_na['date'] >= 1908]
post_1908_rainfall_na_df.head()
```

```python
post_1908_rainfall_na_df.info()
```

```python
# Rename 'date' column to 'year'
post_1908_rainfall_na_df = post_1908_rainfall_na_df.rename(columns={'date': 'year'})
```

```python
# Let's plot rainfall levels in the North America since 1908 (Canada and United States combined)
# Group by year and sum rainfall across all countries
NA_rainfall_per_year = post_1908_rainfall_na_df.groupby('year', as_index=False)['rainfall_mm'].sum()
```

```python

# Plot with Plotly Express
fig = px.scatter(NA_rainfall_per_year, 
              x='year', 
              y='rainfall_mm',
              trendline= 'ols',
              title='Total Rainfall per Year (All Countries)',
              labels={'rainfall_mm': 'Total Rainfall (mm)', 'year': 'Year'})

# Make trendline red
fig.data[1].line.color = "red"
fig.data[1].line.width = 2 

fig.show()
```

```python
# Let's plot rainfall levels in the North America since 1908 (Canada and United States combined)
# Group by year and sum rainfall across all countries
NA_rainfall_per_year = post_1908_rainfall_na_df.groupby('year', as_index=False)['rainfall_mm'].sum()

# Plot with Plotly Express
fig = px.line(NA_rainfall_per_year, 
              x='year', 
              y='rainfall_mm',
              title='Total Rainfall per Year (All Countries)',
              labels={'rainfall_mm': 'Total Rainfall (mm)', 'year': 'Year'})

fig.show()
```

```python
# Prepare data
X = NA_rainfall_per_year['year'].values.reshape(-1, 1)
y = NA_rainfall_per_year['rainfall_mm'].values

# Fit linear regression
model = LinearRegression()
model.fit(X, y)

# Predict values
y_pred = model.predict(X)

# Plot original rainfall as line
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=NA_rainfall_per_year['year'],
    y=y,
    mode='lines',
    name='Rainfall'
))

# Add regression line
fig.add_trace(go.Scatter(
    x=NA_rainfall_per_year['year'],
    y=y_pred,
    mode='lines',
    name='Linear Regression',
    line=dict(dash='dash', color='red')  # dashed red line
))

# Update layout
fig.update_layout(
    title='Total Rainfall per Year (All Countries)',
    xaxis_title='Year',
    yaxis_title='Total Rainfall (mm)'
)

fig.show()
```

```python
# Create a new "country" column based on the prefix of region_code
post_1908_rainfall_na_df['country'] = post_1908_rainfall_na_df['region_code'].str.split('.').str[0]  

# Now group by year + country and sum rainfall
country_rainfall_per_year = (
    post_1908_rainfall_na_df.groupby(['year', 'country'], as_index=False)['rainfall_mm'].mean()
)

print(country_rainfall_per_year.head())
```

```python
# Line plot average rainfall of data
fig = px.line(
    country_rainfall_per_year,
    x="year",
    y="rainfall_mm",
    color="country",   # separate lines for CAN and USA
    title="Average Rainfall per Year (Canada vs USA)",
    labels={"rainfall_mm": "Average Rainfall (mm)", "year": "Year"}
)

fig.show()
```

```python
# Plot a scatter with linear regression
fig = px.scatter(
    country_rainfall_per_year,
    x="year",
    y="rainfall_mm",
    trendline= "ols",
    trendline_scope= "trace",
    color="country",   # separate lines for CAN and USA
    title="Average Rainfall per Year (Canada vs USA)",
    labels={"rainfall_mm": "Average Rainfall (mm)", "year": "Year"}
)

fig.show()
```

```python
# Line plot with linear regression
fig = go.Figure()

for country in country_rainfall_per_year['country'].unique():
    df_country = country_rainfall_per_year[country_rainfall_per_year['country'] == country]
    X = df_country['year'].values.reshape(-1, 1)
    y = df_country['rainfall_mm'].values
    
    # Add the actual rainfall line
    fig.add_trace(go.Scatter(
        x=df_country['year'],
        y=y,
        mode='lines',
        name=f'{country} Rainfall'
    ))
    
    # Fit and add regression line
    model = LinearRegression()
    model.fit(X, y)
    y_pred = model.predict(X)
    
    fig.add_trace(go.Scatter(
        x=df_country['year'],
        y=y_pred,
        mode='lines',
        name=f'{country} Linear Regression',
        line=dict(dash='dash') 
    ))

# Update layout
fig.update_layout(
    title='Average Regional Rainfall per Year (Canada vs USA)',
    xaxis_title='Year',
    yaxis_title='Average Rainfall per Region (mm)'
)

fig.show()
```

Apparently Canada has lower average precipitation numbers than the United States. Canada has large areas that are cold and dry (Arctic). Whereas the United States has very wet regions (Southeast, Pacific Northwest, Gulf Coast, Midwest Thunderstorms) that really play into the national average.
